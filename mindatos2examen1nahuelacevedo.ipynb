```python
from pyspark.sql import SparkSession, functions as f
spark = SparkSession.builder.appName("Nahue").master("local[*]").getOrCreate()
```

```python
list_files = ["billing_monthly", "customers_orgs", "marketing_touches", "nps_surveys", "resources", "support_tickets", "users"]
dic_files = {}
for name in list_files:
  path = "/content/" + name + ".csv"
  dic_files[name] = spark.read.csv(path, header=True, inferSchema=True)
```

```python
for name, dataset in dic_files.items():
  print(f"{name}:")
  dic_files[name].show(1)
  print("-----------------------")
```

```python
for name, dataset in dic_files.items():
  print(name)
  dataset.describe().show()
  print("----------------------")
```

```python
for name, dataset in dic_files.items():
  print(name)
  dic_files[name].select([
    f.count(f.when(f.col(c).isNull(), c)).alias(c) 
    for c in dic_files[name].columns
]).show()
  print("----------------------")
```
